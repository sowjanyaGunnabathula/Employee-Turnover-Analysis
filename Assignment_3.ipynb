```python
import pandas as pd
import numpy as np
```

```python
#Create employee IDs
employee_ids = [
    "EMP001", "EMP002", "EMP003", "EMP004", "EMP005",
    "EMP006", "EMP007", "EMP008", "EMP009", "EMP010",
    "EMP011", "EMP012", "EMP013", "EMP014", "EMP015",
    "EMP016", "EMP017", "EMP018", "EMP019", "EMP020",
    "EMP021", "EMP022", "EMP023", "EMP024", "EMP025",
    "EMP026", "EMP027", "EMP028", "EMP029", "EMP030"
]
```

```python
#Check the employee IDs

print(employee_ids)
print("Total Employees:", len(employee_ids))
```

```python
#Create department names

departments = [
    "HR", "IT", "Finance", "Marketing", "Sales"
]

department_data = np.random.choice(
    departments,
    size=30
)
```

```python
#Create employee status

employee_status = np.random.choice(
    ["Active", "Exited"],
    size=30,
    p=[0.8, 0.2]
)
```

```python
#Create the employee dataset

df = pd.DataFrame({
    "Employee_ID": employee_ids,
    "Department": department_data,
    "Employee_Status": employee_status
})

df.head()
```

```python
#Check the dataset

print("Total Employees:", len(df))

print("\nDepartment-wise count:")
print(df["Department"].value_counts())

print("\nEmployee status:")
print(df["Employee_Status"].value_counts())
```

```python
#Create a training column  p=[0.7, 0.3] – gives each employee a 70% chance of attending and 30% chance of not attending.

df["Training_Attended"] = np.random.choice(
    ["Yes", "No"],
    size=30,
    p=[0.7, 0.3]
)

df.head()
```

```python
#Check the training count

print(df["Training_Attended"].value_counts())
```

```python
#Calculate Training Participation Rate.....Count employees who attended training

trained_employees = (df["Training_Attended"] == "Yes").sum()

total_employees = len(df)

print("Employees who attended training:", trained_employees)
print("Total employees:", total_employees)

```

```python
#Calculate the percentage....2 – tells Python to show two digits after the decimal point.....his displays the percentage symbol after the result.

training_participation_rate = (
    trained_employees / total_employees
) * 100

print("Training Participation Rate:", round(training_participation_rate, 2), "%")
```

```python
#Calculate Absenteeism Rate....Create attendance data
#Let's assume each of our 30 employees had 22 scheduled working days in a month.
#Working_Days = 22: Assigns 22 scheduled working days to each employee.

np.random.randint(0, 5, size=30): Generates random absence days from 0 to 4 for 30 employees.
df["Working_Days"] = 22

df["Absent_Days"] = np.random.randint(0, 5, size=30)

df[["Employee_ID", "Working_Days", "Absent_Days"]].head()

```

```python
#Check total absent days

total_absent_days = df["Absent_Days"].sum()

total_working_days = df["Working_Days"].sum()

print("Total Absent Days:", total_absent_days)
print("Total Available Working Days:", total_working_days)
```

```python
#Calculate Absenteeism Rate

absenteeism_rate = (
    total_absent_days / total_working_days
) * 100

print("Absenteeism Rate:", round(absenteeism_rate, 2), "%")
```

```python
#Employee Turnover Rate
#Employee turnover means the percentage of employees who leave an organization during a specific period.
#Turnover Rate =Number of Employees Who Left\Average Headcount*100
#Average headcount is:Opening Headcount\Closing Headcount%2

opening_headcount = 30
new_joiners = 5
employees_left = 4

closing_headcount = (
    opening_headcount + new_joiners - employees_left
)

average_headcount = (
    opening_headcount + closing_headcount
) / 2

print("Opening Headcount:", opening_headcount)
print("New Joiners:", new_joiners)
print("Employees Left:", employees_left)
print("Closing Headcount:", closing_headcount)
print("Average Headcount:", average_headcount)

```

```python
#Calculate turnover rate


turnover_rate = (
    employees_left / average_headcount
) * 100

print("Employee Turnover Rate:", round(turnover_rate, 2), "%")
```

```python
#Calculate Employee Retention Rate...Employee retention means the organization's ability to keep its existing employees rather than losing them.
#Retention Rate =Opening Employees - Employees Who Left/Opening Employees*100


retained_employees = opening_headcount - employees_left

retention_rate = (
    retained_employees / opening_headcount
) * 100

print("Retained Employees:", retained_employees)
print("Employee Retention Rate:", round(retention_rate, 2), "%")
```

```python
#Calculate Recruitment Conversion Rate...It measures how many candidates are successfully hired compared with the total number of candidates who applied.
#Recruitment Conversion Rate =Candidates Hired\Total Applicants}}\times100

total_applicants = 100
candidates_screened = 40
candidates_interviewed = 20
offers_released = 10
candidates_hired = 8
```

```python
recruitment_conversion_rate = (
    candidates_hired / total_applicants
) * 100

print(
    "Recruitment Conversion Rate:",
    round(recruitment_conversion_rate, 2),
    "%"
)
```

```python
#Create a summary table

metrics_summary = pd.DataFrame({
    "HR Metric": [
        "Employee Turnover Rate",
        "Absenteeism Rate",
        "Training Participation Rate",
        "Recruitment Conversion Rate",
        "Employee Retention Rate"
    ],

    "Calculated Result": [
        round(turnover_rate, 2),
        round(absenteeism_rate, 2),
        round(training_participation_rate, 2),
        round(recruitment_conversion_rate, 2),
        round(retention_rate, 2)
    ],

    "Unit": [
        "%", "%", "%", "%", "%"
    ],

    "HR Interpretation": [
        "Measures the percentage of employees leaving.",
        "Measures employee absence during working days.",
        "Measures employee participation in training.",
        "Measures applicants successfully converted into hires.",
        "Measures the percentage of original employees retained."
    ]
})

metrics_summary

```

```python
#Create a chart of HR performance metrics

import matplotlib.pyplot as plt
```

```python
#create bar chart

plt.figure(figsize=(10, 5))

plt.bar(
    metrics_summary["HR Metric"],
    metrics_summary["Calculated Result"]
)

plt.title("HR Performance Metrics")
plt.xlabel("HR Metrics")
plt.ylabel("Percentage (%)")

plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
```

```python
#Export your summary to Excel
'''pd.ExcelWriter() creates an Excel workbook.
Employee_Data stores your 30 fictional employee records.
Metrics_Summary stores the five calculated HR metrics.
index=False prevents Pandas from adding an unnecessary row-number column.'''

with pd.ExcelWriter("HR_Performance_Metrics.xlsx") as writer:
    df.to_excel(
        writer,
        sheet_name="Employee_Data",
        index=False
    )

    metrics_summary.to_excel(
        writer,
        sheet_name="Metrics_Summary",
        index=False
    )
```

```python
#Create a calculation details table

calculation_details = pd.DataFrame({
    "HR Metric": [
        "Employee Turnover Rate",
        "Absenteeism Rate",
        "Training Participation Rate",
        "Recruitment Conversion Rate",
        "Employee Retention Rate"
    ],

    "Input Values": [
        f"Employees left = {employees_left}; Average headcount = {average_headcount}",
        f"Absent days = {total_absent_days}; Available working days = {total_working_days}",
        f"Employees trained = {trained_employees}; Total employees = {total_employees}",
        f"Candidates hired = {candidates_hired}; Total applicants = {total_applicants}",
        f"Opening employees = {opening_headcount}; Employees left = {employees_left}"
    ],

    "Calculation Formula": [
        "(Employees Left / Average Headcount) * 100",
        "(Absent Days / Available Working Days) * 100",
        "(Employees Trained / Total Employees) * 100",
        "(Candidates Hired / Total Applicants) * 100",
        "((Opening Employees - Employees Left) / Opening Employees) * 100"
    ],

    "Final Result (%)": [
        round(turnover_rate, 2),
        round(absenteeism_rate, 2),
        round(training_participation_rate, 2),
        round(recruitment_conversion_rate, 2),
        round(retention_rate, 2)
    ]
})

calculation_details
```
